```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# from sklearn.preprocessing import StandardScaler, LabelEncoder
# from sklearn.model_selection import train_test_split, GridSearchCV
# from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
# from sklearn.linear_model import LinearRegression
# from sklearn.metrics import r2_score, mean_squared_error
```

```python
# Load the dataset
data = pd.read_csv('student_habits_performance.csv')

print(data.info())

print(data.shape)


```

## Handling missing values by inputing 

```python
# Confirming that there is no missing values
data['parental_education_level'].isnull().sum()
```

```python
# Checking for Object columns and removing white spaces
```

```python
object_cols = ['student_id',
                    'gender',
                    'part_time_job',
                    'diet_quality',
                    'parental_education_level',
                    'internet_quality',
                    'extracurricular_participation'
]
# Remove leading/trailing whitespaces from object columns
print('Leading and trailing whitespaces removed from all string columns in the dataset.')
for col in object_cols:
    data[col] = data[col].str.strip()
    print(col)
```

## Visualization of the data

```python


# Set up the plot style
sns.set_style("whitegrid")

# Create visualizations for numerical features vs. exam_score (Scatter Plots)

# 1. Study Hours per Day vs. Exam Score
plt.figure(figsize=(10, 6))
sns.scatterplot(x='study_hours_per_day', y='exam_score', hue='gender',data=data, alpha=0.7)
plt.title('Exam Score vs. Study Hours per Day')
plt.xlabel('Study Hours per Day')
plt.ylabel('Exam Score')
plt.show()

# 2. Attendance Percentage vs. Exam Score
plt.figure(figsize=(10, 6))
sns.scatterplot(x='attendance_percentage', y='exam_score', hue = 'gender', data=data, alpha=0.7)
plt.title('Exam Score vs. Attendance Percentage')
plt.xlabel('Attendance Percentage')
plt.ylabel('Exam Score')
plt.savefig('exam_score_vs_attendance.png')
plt.show()

# 3. Social Media Hours vs. Exam Score
plt.figure(figsize=(10, 6))
sns.scatterplot(x='social_media_hours', y='exam_score', data=data, alpha=0.7)
plt.title('Exam Score vs. Social Media Hours')
plt.xlabel('Social Media Hours')
plt.ylabel('Exam Score')
plt.savefig('exam_score_vs_social_media.png')
plt.show()

# 4. Sleep Hours vs. Exam Score
plt.figure(figsize=(10, 6))
sns.scatterplot(x='sleep_hours', y='exam_score', data=data, alpha=0.7)
plt.title('Exam Score vs. Sleep Hours')
plt.xlabel('Sleep Hours')
plt.ylabel('Exam Score')
plt.savefig('exam_score_vs_sleep_hours.png')
plt.show()


# Visualizations for categorical features vs. exam_score (Box Plots)

# 5. Gender vs. Exam Score
plt.figure(figsize=(8, 6))
sns.boxplot(x='gender', y='exam_score', data=data)
plt.title('Exam Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Exam Score')
plt.savefig('exam_score_by_gender.png')
plt.show()

# 6. Parental Education Level vs. Exam Score
plt.figure(figsize=(12, 7))
# Order education categories  for better visualization
ordered_education = ['No Education', 'High School', 'Bachelor', 'Master', 'PhD']
sns.boxplot(x='parental_education_level', y='exam_score', hue='gender', data=data, order=ordered_education)
plt.title('Exam Score Distribution by Parental Education Level')
plt.xlabel('Parental Education Level')
plt.ylabel('Exam Score')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('exam_score_by_parental_education.png')
plt.show()

# 7. Diet Quality vs. Exam Score
plt.figure(figsize=(8, 6))
ordered_diet = ['Poor', 'Fair', 'Good']
sns.boxplot(x='diet_quality', y='exam_score', color='green',data=data, order=ordered_diet)
plt.title('Exam Score Distribution by Diet Quality')
plt.xlabel('Diet Quality')
plt.ylabel('Exam Score')
plt.savefig('exam_score_by_diet_quality.png')
plt.show()
```
